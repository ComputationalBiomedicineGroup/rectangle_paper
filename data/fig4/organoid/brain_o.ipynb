# Rectangle Deconvolution of brain organoid data (BrainO)

This notebook demonstrates cell-type deconvolution using Rectangle.

```python
import rectanglepy as rp
import pandas as pd
from anndata import AnnData, read_h5ad
from scipy.sparse import csr_matrix
```

## Day013 top 10 unknown-error genes per protocol/cell line


```python
err = pd.read_csv("braino_rectangle_err.csv", index_col=0)

# Parse sample IDs: replicate-cell_line-protocol-day-...
sample_meta = err.index.to_series().str.split("-", expand=True)

sample_meta = sample_meta.iloc[:, :4].copy()
sample_meta.columns = ["replicate", "cell_line", "protocol", "day"]

# Keep first time point only (day013)
day013_mask = sample_meta["day"] == "day013"
err_day013 = err.loc[day013_mask].copy()
meta_day013 = sample_meta.loc[day013_mask, ["protocol", "cell_line"]].copy()

# Mean error per gene across replicates for each protocol and cell line
err_mean = (
    err_day013
    .assign(
        protocol=meta_day013["protocol"].values,
        cell_line=meta_day013["cell_line"].values,
    )
    .groupby(["protocol", "cell_line"], as_index=True)
    .mean(numeric_only=True)
)

# Extract top 10 genes with highest mean error in each group
top10_tables = []
for (protocol, cell_line), row in err_mean.iterrows():
    top10 = row.nlargest(10).rename("mean_error").reset_index()
    top10.columns = ["gene", "mean_error"]
    top10.insert(0, "cell_line", cell_line)
    top10.insert(0, "protocol", protocol)
    top10.insert(2, "rank", range(1, len(top10) + 1))
    top10_tables.append(top10)

braino_day013_top10_err = pd.concat(top10_tables, ignore_index=True)
braino_day013_top10_err = braino_day013_top10_err.sort_values(["protocol", "cell_line", "rank"])

out_file = "braino_day013_top10_mean_error_by_protocol_cellline.csv"
braino_day013_top10_err.to_csv(out_file, index=False)

print(f"Saved: {out_file}")
braino_day013_top10_err
```
